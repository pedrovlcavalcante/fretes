```python
import pandas as pd
```

```python
df = pd.read_excel("braspress_jan_jul.xlsx").drop(columns=["Unnamed: 0", "NUM CONHECIMENTO", "NUM PEDIDO", "CNPJ REMETENTE", "REMETENTE", 
                                                           "TIPO EMISSÃO", "UF ORIGEM", "CNPJ DESTINATÁRIO", "DESTINATARIO", "DATA EMISSÃO", 
                                                           "CIDADE ORIGEM", "UF ORIGEM"])
df.head()
```

# Limpeza de dados

```python
df.info()
```

```python
df["NOTA FISCAL"] = df["NOTA FISCAL"].apply(lambda x: x.replace(",", ""))
```

```python
df["PREVISÃO DE ENTREGA"] = df["PREVISÃO DE ENTREGA"].astype("datetime64[s]")
df["PREVISÃO DE ENTREGA ORIGINAL"] = df["PREVISÃO DE ENTREGA ORIGINAL"].astype("datetime64[s]")
df["DATA OCORRÊNCIA"] = df["DATA OCORRÊNCIA"].astype("datetime64[s]", errors="ignore")
```

```python
df.fillna("-", inplace=True)
```

```python
df["NOTA FISCAL"]=df["NOTA FISCAL"].astype("int64")
```

```python
df.info()
```

```python
df.loc[df["NOTA FISCAL"] == 9586]
```

```python
faturados = pd.read_excel('faturados_jan_jul.xls').dropna().reset_index(drop=True).drop(columns=['DESTINATARIO', 'Nome Fantasia', 'CNPJ DESTINATÁRIO', 'Faturado'])
faturados["NOTA FISCAL"] = faturados["NOTA FISCAL"].astype("int64")
faturados.head()
```

```python
faturados.info()
```

```python
combina = pd.merge(df, faturados, on="NOTA FISCAL")
dados = combina[["Pedido", "NOTA FISCAL", "MODAL", "CIDADE DESTINO_x", "UF DESTINO_x", "PESO", "VOLUME", "VALOR MERCADORIA", "Valor total do pedido", "VALOR TOTAL FRETE"]]
dados["VALOR MERCADORIA"] = dados['VALOR MERCADORIA'].apply(lambda x: x.replace('.','').replace(',','.')).astype(float)
dados['VALOR TOTAL FRETE'] = dados['VALOR TOTAL FRETE'].apply(lambda x: x.replace('.','').replace(',','.')).astype(float)
dados['PESO'] = dados['PESO'].apply(lambda x: x.replace('.','').replace(',','.')).astype(float)
dados['PESO']
```

```python
dados.info()
```

```python
dados['MODAL'] = dados['MODAL'].apply(lambda x: 0 if x == "RODO" else 1)
dados['MODAL']
```

```python
dados_visualizacao = dados.drop(['Pedido', 'NOTA FISCAL', 'Valor total do pedido'], axis=1)
dados_visualizacao.to_csv('dados_visualizacao_braspress.csv', index=False)
```

```python
dados_sem_cidade = dados.drop(['CIDADE DESTINO_x', 'Pedido', 'NOTA FISCAL', 'Valor total do pedido'], axis=1)
dados_sem_cidade.head()
```

```python
dados_sem_cidade.to_csv('dados_treino_braspress_1_uf.csv', index=False)
```
