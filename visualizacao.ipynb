```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('dados_visualizacao_braspress.csv')
df.head()
```

```python
df.describe()
```

```python
matriz_corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))

# 3. Cria o mapa de calor com o Seaborn
sns.heatmap(matriz_corr, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)

# 4. Exibe o gráfico
plt.show()
```

```python
grupos = df.groupby(['UF DESTINO_x'])['VALOR TOTAL FRETE'].mean().sort_values(ascending=False)
grupos[0:5].plot(kind='bar')
```

```python
modais = df.groupby(['MODAL'])['VALOR TOTAL FRETE'].mean().sort_values(ascending=False)
modais.plot(kind='bar', legend=True)
```
